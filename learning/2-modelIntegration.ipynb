Integration with OpenAI


```python
from langchain.agents import create_agent
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
response = model.invoke(
    "What is today's weather in Chennai?"
)
response.content
```

Google GEMINI


```python
from langchain.chat_models import init_chat_model

google_model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = google_model.invoke(
    "Vanakkam in English?"
)   
response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Hi How are you?")
response.content
```

GROQ Model 

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke(
    "Vanakkam in English?"
)
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Hi How are you?")
response.content
```

```python
for chunk in google_model.stream("Give a 200 word review about the movie - Obsession"):
    print(chunk.text)
```

Batching 

```python
responses = google_model.batch([
    "What is Gen AI?",
    "What is Forward Deployed Engineer?",
    "Cost of Maruti Suzuki Ciaz"
])

for response in responses:
    print(response)
```
